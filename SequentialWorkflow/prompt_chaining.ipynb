```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()

model = ChatOpenAI()
```

```python
class BlogState(TypedDict):
    title : str
    outline : str
    content : str
```

```python
def create_outline(state: BlogState) -> BlogState:
    title = state['title']
    prompt = f'generate a detailed outline for a blog on the topic - {title}'
    outline = model.invoke(prompt).content
    state['outline'] = outline
    return state
```

```python
def create_blog(state: BlogState) -> BlogState:
    title = state['title']
    outline = state['outline']

    prompt = f'craete a detailed blog on the title {title} using the following outline {outline}' 

    content = model.invoke(prompt).content

    state['content'] = content

    return state
```

```python
llm_graph = StateGraph(BlogState)

llm_graph.add_node('create_outline',create_outline)
llm_graph.add_node('create_blog',create_blog)

llm_graph.add_edge(START,'create_outline')
llm_graph.add_edge('create_outline','create_blog')
llm_graph.add_edge('create_blog',END)

workflow = llm_graph.compile()
```

```python
initial_state = {'title': "Dream 11"}

final_state = workflow.invoke(initial_state)

# print(final_state)
print(final_state["content"])
```
