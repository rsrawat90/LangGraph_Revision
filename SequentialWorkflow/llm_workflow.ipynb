```python
from langgraph.graph import StateGraph,START,END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()

model = ChatOpenAI()
```

```python
# define state

class llm_State(TypedDict):
    question: str
    answer: str
```

```python
# define node
def llm_qa(state: llm_State) -> llm_State:
    # extract the question from state
    question = state['question']
    
    # draft a promt
    prompt = f"answer the following question {question}"

    # call LLM
    answer = model.invoke(prompt).content
     
    # save the answer 
    state['answer'] = answer

    return state
```

```python
# define graph
llm_graph = StateGraph(llm_State)

# define node
llm_graph.add_node('llm_qa',llm_qa)
llm_graph.add_edge(START,'llm_qa')
llm_graph.add_edge('llm_qa',END)
```

```python
# compile graph

workflow = llm_graph.compile()
```

```python
# execute the workflow

initial_state = {'question':"What is the capital of India"} 
output_state = workflow.invoke(initial_state)
print(output_state['answer'])
```
